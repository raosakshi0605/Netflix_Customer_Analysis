# Netflix Dummy Dataset Project

```python
# Import libraries

import pandas as pd
import numpy as np
```

```python
# Load the dataset
data_frame = pd.read_excel('Netflix_Unclean_Project.xlsx')
print(data_frame)
```

```python
# Information about the dataset

print(f'No. of rows and columns: {data_frame.shape}')
print(f'Column names: {data_frame.columns}')
print(f'Info: {data_frame.info()}')
print(f'Statistical summary: {data_frame.describe()}')
```

```python
# Null Values

print(f'Null Values: {data_frame.isnull().sum()}') #.sum() because we want to see all null data together
```

```python
# Find Duplicate data and remove duplicates

print(f'Duplicate data: {data_frame.duplicated().sum()}')

# Dropping it 
# inplace = True means that the changes will be made in the original dataframe itself, and it will not return a new dataframe.
data_frame = data_frame.drop_duplicates(inplace=True)
```

```python
# Dirty values - Null, N/A , blank , spaces etc 
# This means we replace the stated values by np.nan (which is representation of null values in numpy)
data_frame = data_frame.replace(['N/A', 'NA', 'na', 'n/a', ' ',''],np.nan)

# Now we'll use the isnull() function again to find the changes innull values in the dataset.

data_frame.isnull().sum()
```

```python
# Extra Space remove and Standardize the text data - Title Case
# .apply() - applies the function to all the columns in one go.
# Here we are using lambda function to apply strip() to x where x is an object i.e string/textual colum or return x otherwise.

data_frame = data_frame.apply(lambda x: x.str.strip().title() if x.dtype == 'object' else x)

print(data_frame)
```

```python
# Fixing numeric columns

numeric_columns = ['Age', 'Watch_Hours', 'Monthly_Fee', 'Movies_Watched', 'Episodes_Watched', 'Avg_Rating', 'Monthly_Sessions', 'Download_Count']

# error = 'coerce' means that if there are any values that cannot be converted to numeric or blank spaces etc, they will be replaced with NaN (Not a Number) instead of raising an error.
for column in numeric_columns :
    data_frame[column] = pd.to_numeric(data_frame[column], errors = 'coerce')

data_frame[numeric_columns].info() # to check the changes made in the numeric columns

data_frame.describe()
```

```python
# Fixing negative and blank values in numeric columns

data_frame = data_frame[(data_frame['Age'] >= 14) & (data_frame['Age'] <= 110)]

data_frame = data_frame[(data_frame['Watch_Hours']>= 0)]

data_frame = data_frame[(data_frame['Monthly_Fee']>= 0)]
```

```python
# Fixing date columns
 
data_frame['Join_Date'] = pd.to_datetime(data_frame['Join_Date'], errors = 'coerce')
data_frame['Last_Watch_Date'] = pd.to_datetime(data_frame['Last_Watch_Date'], errors = 'coerce')
```

```python
# Missing values handle

# Blank/missing data in numeric columns is replaced by their mean (using inplace method using {key:value} pairs)
data_frame.fillna({'Age' : data_frame['Age'].mean()}, inplace = True)

data_frame.fillna({'Watch_Hours' : data_frame['Watch_Hours'].mean()}, inplace = True)

data_frame.fillna({'Monthly_Fee' : data_frame['Monthly_Fee'].mean()}, inplace = True)


# Blank/missing data in string columns is replaced by 'Unknown'ArithmeticError
data_frame.fillna({'Payment_Method':'Unknown'}, inplace = True)

data_frame.fillna({'Device':'Unknown'}, inplace = True)

data_frame.fillna({'Genre':'Unknown'}, inplace = True)

data_frame.fillna({'Country':'Unknown'}, inplace = True)
```

```python
# Rechecking

data_frame.describe()
```

```python
# Adding some new Columns 

# Customer Revenue
data_frame['Customer_Revenue'] = (data_frame['Monthly_Fee'] * data_frame['Monthly_Sessions'])

# Engagement Score
data_frame['Engagement_Score'] = (data_frame['Watch_Hours'] + data_frame['Monthly_Sessions'] + data_frame['Download_Count'])

# Watch Category 
bins = [0,25,50,100, np.inf] # np.inf means any value accepted above 100 for very high label
labels = ['Low', 'Medium', 'High', 'Very High'] # these are the labels we have to put on the above values of bins

# include_lowest = True because by deafault bins does not include the lowest i.e left value so to make 0 included we have to make it true
data_frame['Watch_Category'] = pd.cut(data_frame['Watch_Hours'], bins = bins, labels = labels, include_lowest = True)

# Active Flags
# If active = 1 , otherwise 0

data_frame['Active_Flag'] = np.where(data_frame['Account_Status'] == 'Active',1 , 0)
```

```python
print(data_frame.head())

data_frame['Watch_Category'].value_counts()
```

```python
# Final Check

data_frame.info()

data_frame.describe()

print(f'Null Values: {data_frame.isnull().sum()}')

data_frame = data_frame.drop_duplicates(inplace=True)
print(f'Duplicate values: {data_frame.duplicated().sum()}')

```
